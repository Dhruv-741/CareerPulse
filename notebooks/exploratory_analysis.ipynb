# CareerPulse — Exploratory Data Analysis & Feature Engineering

**Project:** CareerPulse — AI-Powered Job Market Intelligence and Career Readiness Analyzer  
**Objective:** Academic exploration of job posting patterns, skill distributions, experience levels, and preparation for supervised machine learning models.  
**Dataset:** LinkedIn Job Postings (2023–2024 schema) with ~15,000 raw postings.

```python
import os
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure project root is in path
sys.path.append('..')
from src.data_loader import load_raw_data, load_processed_data
from src.preprocessing import clean_dataframe, split_data
```

## 1. Raw Data Ingestion & Quality Audit

```python
raw_jobs, raw_companies, raw_skills = load_raw_data()
print(f"Raw Postings: {raw_jobs.shape}")
print(f"Raw Companies: {raw_companies.shape}")
print(f"Raw Skills: {raw_skills.shape}")
raw_jobs.head(3)
```

## 2. Missing Value & Duplication Analysis

```python
missing_stats = raw_jobs.isnull().sum()
missing_pct = (missing_stats / len(raw_jobs) * 100).round(2)
pd.DataFrame({'Missing Count': missing_stats, 'Percentage (%)': missing_pct})

duplicates = raw_jobs.duplicated(subset=['job_id']).sum()
print(f"Exact Duplicate job_id rows: {duplicates}")
```

## 3. Data Cleaning & Normalization

```python
clean_df = clean_dataframe(raw_jobs)
print(f"Cleaned DataFrame shape: {clean_df.shape}")
clean_df['formatted_experience_level'].value_counts()
```

## 4. Exploratory Visualizations

```python
plt.figure(figsize=(10, 4))
sns.countplot(data=clean_df, y='domain', order=clean_df['domain'].value_counts().index, palette='crest')
plt.title('Job Postings by Domain')
plt.xlabel('Count')
plt.ylabel('Domain')
plt.show()
```

```python
plt.figure(figsize=(10, 4))
sns.countplot(data=clean_df, x='formatted_experience_level', order=['Entry level', 'Associate', 'Mid-Senior level', 'Senior', 'Director', 'Executive'], palette='viridis')
plt.title('Experience Level Class Distribution')
plt.xlabel('Experience Level')
plt.ylabel('Count')
plt.xticks(rotation=15)
plt.show()
```

## 5. Leakage-Safe Chronological Split Verification

```python
train_df, val_df, test_df = split_data(clean_df, test_size=0.15, val_size=0.15)
print(f"Train set: {len(train_df)} ({len(train_df)/len(clean_df):.1%})")
print(f"Val set:   {len(val_df)} ({len(val_df)/len(clean_df):.1%})")
print(f"Test set:  {len(test_df)} ({len(test_df)/len(clean_df):.1%})")
```
